Import sql3 and import pandas

```python
import sqlite3
import pandas as pd
```

Create table from csv using pandas

```python
def add_data_to_database(input_data):
  try:
    df = pd.read_csv(input_data)
    conn = sqlite3.connect('company_demon_data.db')
    df.to_sql('safety', con=conn, index=False)
    print("Information added sucessfully")
  except Exception as e:
    print(e)


data = "/content/drive/MyDrive/A_SQL_Essentials_For_Data_Analysis_Datasets/company_safety.csv"
add_data_to_database(data)
```

connection to db; create cursor; cursor.execute(SELECT FROM); results;  

```python
conn = sqlite3.connect('company_demon_data.db')
cursor = conn.cursor()
cursor.execute("""
SELECT * FROM safety
""")

results = cursor.fetchall()

for row in results:
  print(row)
```
